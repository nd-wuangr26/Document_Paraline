```python
import cv2
import matplotlib.pyplot as plt

def show_image_cv2(img_path):
    """
    Hiển thị ảnh từ đường dẫn bằng OpenCV trong notebook.
    """
    # Đọc ảnh (cv2 đọc mặc định là BGR)
    img = cv2.imread(img_path)

    # Chuyển sang RGB để hiển thị đúng màu
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Hiển thị bằng matplotlib
    plt.figure(figsize=(8,6))
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.show()
```

```python
from paddleocr import PaddleOCR
ocr = PaddleOCR(
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
    lang = "vi")
```

```python
result = ocr.predict(
    input="/content/Thực_đơn_viết_tay_các_món_cúng_đưa_Tết_2018.jpg")

# Visualize the results and save the JSON results
for res in result:

    res.save_to_img("output")
    res.save_to_json("output")
```

```python
import json

# Giả sử bạn đã lưu JSON này vào file 'ocr_result.json'
with open("/content/output/Thực_đơn_viết_tay_các_món_cúng_đưa_Tết_2018_res.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# In từng dòng văn bản
for line in data["rec_texts"]:
    print(line)
```
