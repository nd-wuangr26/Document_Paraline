```python
try:
 from PIL import Image
except ImportError:
 import Image
import cv2
import pytesseract
```

```python
img_cv = cv2.imread(r'/content/Sentences-with-Document-Document-in-a-Sentence-in-English-Sentences-for-Document.webp')
# By default OpenCV stores images in BGR format and since pytesseract assumes RGB format,
# we need to convert from BGR to RGB format/mode:
#d = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
```

```python
import matplotlib.pyplot as plt


# 1. Đọc ảnh (đường dẫn ví dụ: 'sample.jpg')
img_cv = cv2.imread('/content/Sentences-with-Document-Document-in-a-Sentence-in-English-Sentences-for-Document.webp')

# 2. (Tuỳ chọn) chuyển sang RGB vì OpenCV đọc ảnh theo BGR
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)

# 3. Hiển thị ảnh trên Jupyter
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.axis('off')         # Ẩn trục
plt.show()
```

```python
from pprint import pprint
custom_oem_psm_config = r'--oem 3 --psm 1'
pprint(pytesseract.image_to_string(img_cv, config=custom_oem_psm_config, lang = 'eng'))
```
