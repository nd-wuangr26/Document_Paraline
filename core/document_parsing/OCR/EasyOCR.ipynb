```python
import easyocr
import cv2
from google.colab.patches import cv2_imshow   # Dùng hàm hiển thị cho Colab

# Khởi tạo EasyOCR (ngôn ngữ tiếng Việt)
reader = easyocr.Reader(['vi','eng'])




```

```python
img_path = '/content/Screen-Shot-2017-11-.png'
# Đọc văn bản
result = reader.readtext(img_path)

# Đọc ảnh
mat = cv2.imread(img_path)

# Lấy các box, text, score
boxes  = [line[0] for line in result]
texts  = [line[1] for line in result]
scores = [line[2] for line in result]

# Vẽ khung cho từng vùng phát hiện
for box in boxes:
    top_left     = (int(box[0][0]), int(box[0][1]))
    bottom_right = (int(box[2][0]), int(box[2][1]))
    cv2.rectangle(mat, top_left, bottom_right, (0, 255, 0), 2)
# Hiển thị ảnh trong Colab
cv2_imshow(mat)
```

```python
# In kết quả OCR
from pprint import pprint
pprint(texts)
```
