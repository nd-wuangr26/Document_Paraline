```python
from docling.document_converter import DocumentConverter
from docling_core.transforms.chunker import HierarchicalChunker
```

```python
# Chuyển đổi tài liệu sang DoclingDocument
source = "/home/quang/My_Project/Paraline/paraline/document/Huong-dan-doc-va-phan-tich-BCTC-NHANH.pdf"
doc = DocumentConverter().convert(source=source).document
```

```python
# Instantiate HierarchicalChunker, có thể bật/tắt merge_list_items
chunker = HierarchicalChunker(merge_list_items=True)  # hoặc False
```

```python
# Lấy các chunk
chunks_iter = chunker.chunk(doc)
```

```python
chunks = list(chunks_iter)
for i, ch in enumerate(chunks[:5]):
    print(f"=== Chunk {i} ===")
    print("Text:", ch.text[:200], "…")
    print("Metadata:", ch.meta)  # meta bao gồm headings, captions, page, etc.
    print()
```
