```python
from dotenv import load_dotenv
import os
load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

class ChainOfThoughtDemo:
    def __init__(self):
        self.llm = ChatOpenAI(model="gpt-4o-mini", api_key=OPENAI_API_KEY, temperature=0)
    
    def zero_shot_cot(self, problem):
        """Zero-shot CoT: Thêm câu hướng dẫn để suy nghĩ từng bước"""
        prompt = f"""
        {problem}
        
        Let's think step by step to solve this problem.
        """
        return self.llm.invoke(prompt).content
    
    def few_shot_cot(self, problem):
        """Few-shot CoT: Cung cấp ví dụ với cách suy nghĩ từng bước"""
        prompt_template = PromptTemplate.from_template("""
        Here are some examples of step-by-step problem solving:
        
        Problem: What is 15 + 27 - 8?
        Solution:
        Step 1: First, add 15 + 27 = 42
        Step 2: Then subtract 8 from 42: 42 - 8 = 34
        Answer: 34
        
        Problem: If a box contains 12 red balls and 8 blue balls, what percentage are red?
        Solution:
        Step 1: Calculate total balls: 12 + 8 = 20 balls
        Step 2: Calculate percentage of red balls: (12 ÷ 20) × 100 = 60%
        Answer: 60%
        
        Problem: {problem}
        Solution:
        """)
        
        formatted_prompt = prompt_template.format(problem=problem)
        return self.llm.invoke(formatted_prompt).content
    
    def auto_cot_demo(self, problem):
        """Auto-CoT: Tự động tạo reasoning chains"""
        auto_prompt = f"""
        Generate a step-by-step solution for the following problem.
        Break down your reasoning into clear, logical steps.
        
        Problem: {problem}
        
        Think about this systematically:
        1. What information do I have?
        2. What do I need to find?
        3. What steps should I follow?
        4. How can I verify my answer?
        
        Solution:
        """
        return self.llm.invoke(auto_prompt).content
```

```python
# Sử dụng demo
demo = ChainOfThoughtDemo()

problem = "A store has 45 items. If 20% are sold in the morning and 15 more are sold in the afternoon, how many items are left?"

print("=== ZERO-SHOT CHAIN OF THOUGHT ===")
result_zero = demo.zero_shot_cot(problem)
print(result_zero)

print("\n=== FEW-SHOT CHAIN OF THOUGHT ===")
result_few = demo.few_shot_cot(problem)
print(result_few)

print("\n=== AUTO CHAIN OF THOUGHT ===")
result_auto = demo.auto_cot_demo(problem)
print(result_auto)
```
