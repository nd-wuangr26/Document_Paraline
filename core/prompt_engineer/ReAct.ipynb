```python
import re

class SimpleReActAgent:
    """Agent ReAct đơn giản với các tools cơ bản"""
    
    def __init__(self):
        # Định nghĩa các tools có sẵn
        self.tools = {
            "calculate": self.calculate_tool,
            "search": self.search_tool,
            "translate": self.translate_tool
        }
    
    def calculate_tool(self, expression):
        """Tool tính toán đơn giản"""
        try:
            # Chỉ cho phép các phép tính cơ bản an toàn
            allowed_chars = set('0123456789+-*/.() ')
            if all(c in allowed_chars for c in expression):
                result = eval(expression)
                return f"Kết quả: {result}"
            else:
                return "Lỗi: Biểu thức chứa ký tự không hợp lệ"
        except Exception as e:
            return f"Lỗi tính toán: {str(e)}"
    
    def search_tool(self, query):
        """Tool tìm kiếm thông tin (giả lập)"""
        knowledge_base = {
            "python": "Python là ngôn ngữ lập trình cấp cao, được tạo bởi Guido van Rossum năm 1991",
            "ai": "AI (Trí tuệ nhân tạo) là khả năng máy tính mô phỏng trí tuệ con người",
            "react": "ReAct là framework kết hợp Reasoning (suy luận) và Acting (hành động)",
            "vietnam": "Việt Nam là quốc gia Đông Nam Á với dân số khoảng 98 triệu người",
            "langchain": "LangChain là framework phát triển ứng dụng AI sử dụng mô hình ngôn ngữ"
        }
        
        query_lower = query.lower()
        for key, value in knowledge_base.items():
            if key in query_lower:
                return f"Thông tin: {value}"
        
        return f"Không tìm thấy thông tin về: {query}"
    
    def translate_tool(self, text):
        """Tool dịch thuật đơn giản (English <-> Vietnamese)"""
        translations = {
            "hello": "xin chào",
            "thank you": "cảm ơn",
            "good morning": "chào buổi sáng",
            "how are you": "bạn khỏe không",
            "goodbye": "tạm biệt",
            "xin chào": "hello",
            "cảm ơn": "thank you",
            "chào buổi sáng": "good morning"
        }
        
        text_lower = text.lower()
        for key, value in translations.items():
            if key in text_lower:
                return f"Dịch: {value}"
        
        return f"Không thể dịch: {text}"
    
    def react_solve(self, problem, max_steps=5):
        """
        Giải quyết vấn đề sử dụng ReAct pattern
        
        Cấu trúc:
        1. Thought: Suy nghĩ về vấn đề
        2. Action: Thực hiện hành động cụ thể
        3. Observation: Quan sát kết quả
        4. Lặp lại cho đến khi có đủ thông tin
        5. Final Answer: Đưa ra câu trả lời cuối cùng
        """
        
        print("=" * 60)
        print("ReAct Agent đang giải quyết vấn đề...")
        print("=" * 60)
        print(f"Câu hỏi: {problem}")
        print("\nQuá trình ReAct:")
        
        step = 1
        context = f"Vấn đề cần giải quyết: {problem}"
        
        # Các kịch bản phổ biến và cách xử lý
        if any(op in problem for op in ['+', '-', '*', '/', 'tính', 'cộng', 'trừ', 'nhân', 'chia']):
            return self._solve_math_problem(problem, step)
        elif any(word in problem.lower() for word in ['thông tin', 'tìm kiếm', 'là gì', 'what is']):
            return self._solve_info_problem(problem, step)
        elif any(word in problem.lower() for word in ['dịch', 'translate']):
            return self._solve_translation_problem(problem, step)
        else:
            return self._solve_general_problem(problem, step)
    
    def _solve_math_problem(self, problem, step):
        """Giải quyết bài toán"""
        print(f"\nBước {step}:")
        print("Thought: Đây là bài toán cần tính toán. Tôi sẽ trích xuất biểu thức và sử dụng tool calculate.")
        
        # Trích xuất biểu thức từ câu hỏi
        import re
        numbers_and_ops = re.findall(r'[\d+\-*/().]+', problem)
        if numbers_and_ops:
            expression = ''.join(numbers_and_ops)
        else:
            # Thử tìm số và phép toán bằng từ
            expression = problem.replace('cộng', '+').replace('trừ', '-').replace('nhân', '*').replace('chia', '/')
            expression = re.findall(r'[\d+\-*/().]+', expression)
            expression = ''.join(expression) if expression else "0"
        
        print(f"Action: calculate[{expression}]")
        result = self.calculate_tool(expression)
        print(f"Observation: {result}")
        
        print(f"\nBước {step + 1}:")
        print("Thought: Đã có kết quả tính toán. Tôi có thể đưa ra câu trả lời cuối cùng.")
        print(f"Final Answer: {result}")
        
        return result
    
    def _solve_info_problem(self, problem, step):
        """Giải quyết vấn đề tìm kiếm thông tin"""
        print(f"\nBước {step}:")
        print("Thought: Đây là câu hỏi tìm kiếm thông tin. Tôi sẽ sử dụng tool search để tìm kiếm.")
        
        # Trích xuất từ khóa tìm kiếm
        keywords = ['python', 'ai', 'react', 'vietnam', 'langchain']
        search_term = problem
        for keyword in keywords:
            if keyword.lower() in problem.lower():
                search_term = keyword
                break
        
        print(f"Action: search[{search_term}]")
        result = self.search_tool(search_term)
        print(f"Observation: {result}")
        
        print(f"\nBước {step + 1}:")
        print("Thought: Đã tìm thấy thông tin liên quan. Tôi có thể trả lời câu hỏi.")
        print(f"Final Answer: {result}")
        
        return result
    
    def _solve_translation_problem(self, problem, step):
        """Giải quyết vấn đề dịch thuật"""
        print(f"\nBước {step}:")
        print("Thought: Đây là yêu cầu dịch thuật. Tôi sẽ trích xuất văn bản cần dịch và sử dụng tool translate.")
        
        # Trích xuất văn bản cần dịch (đơn giản)
        text_to_translate = problem.replace('dịch', '').replace('translate', '').strip('"\' ')
        
        print(f"Action: translate[{text_to_translate}]")
        result = self.translate_tool(text_to_translate)
        print(f"Observation: {result}")
        
        print(f"\nBước {step + 1}:")
        print("Thought: Đã có kết quả dịch thuật. Tôi có thể đưa ra câu trả lời.")
        print(f"Final Answer: {result}")
        
        return result
    
    def _solve_general_problem(self, problem, step):
        """Giải quyết vấn đề tổng quát"""
        print(f"\nBước {step}:")
        print("Thought: Đây là câu hỏi tổng quát. Tôi sẽ thử tìm kiếm thông tin trước.")
        
        print(f"Action: search[{problem}]")
        result = self.search_tool(problem)
        print(f"Observation: {result}")
        
        print(f"\nBước {step + 1}:")
        print("Thought: Đã tìm kiếm thông tin. Dựa trên kết quả, tôi sẽ đưa ra câu trả lời.")
        print(f"Final Answer: {result}")
        
        return result
```

```python
def demo_react_examples():
    """Chạy các ví dụ demo ReAct"""
    
    agent = SimpleReActAgent()
    
    # Danh sách các câu hỏi demo
    test_questions = [
        "Tính 15 + 27 * 2",
        "Python là gì?",
        "Dịch 'hello' sang tiếng Việt",
        "AI có nghĩa là gì?",
        "Tính diện tích hình tròn có bán kính 5 (sử dụng 3.14)"
    ]
    
    print("\nDanh sách tools có sẵn:")
    print("   - calculate: Thực hiện phép tính")
    print("   - search: Tìm kiếm thông tin")
    print("   - translate: Dịch thuật đơn giản")
    
    for i, question in enumerate(test_questions, 1):
        print(f"\n\nVÍ DỤ {i}:")
        agent.react_solve(question)
        print("=" * 60)
```

```python
import os
import re
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

class LangChainReActAgent:
    def __init__(self, api_key=None):
        # Thiết lập client
        os.environ["OPENAI_API_KEY"] = api_key or os.environ.get("OPENAI_API_KEY", "")
        self.llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)
        # Định nghĩa tools
        self.tools = {
            "calculate": self._calculate,
        }

    def _calculate(self, expr: str) -> str:
        """Tool tính toán an toàn"""
        try:
            allowed = set("0123456789+-*/.() ")
            if not all(c in allowed for c in expr):
                return "Error: Invalid characters"
            return str(eval(expr))
        except Exception as e:
            return f"Error: {e}"

    def solve(self, question: str, max_steps: int = 4) -> str:
        """Giải bài toán theo ReAct pattern"""
        system_prompt = PromptTemplate.from_template(
            """Bạn là trợ lý AI sử dụng ReAct (Reasoning + Acting) để giải toán.
                Format phải tuân theo:
                Thought: [suy nghĩ]
                Action: [tool_name[parameters]]
                Observation: [kết quả]
                ... (lặp lại)
                Final Answer: [kết quả cuối cùng]

                Giải toán sau:
                {question}
                """)
        messages = [
            {"role": "system", "content": system_prompt.format(question=question)},
        ]
        for step in range(max_steps):
            resp = self.llm.invoke(messages).content
            messages.append({"role": "assistant", "content": resp})
            if "Final Answer:" in resp:
                return resp
            m = re.search(r"Action:\s*(\w+)\[([^\]]+)\]", resp)
            if not m:
                messages.append({"role": "user", "content": "Tiếp tục theo ReAct format."})
                continue
            tool, param = m.group(1), m.group(2)
            obs = self.tools.get(tool, lambda _: "Unknown tool")(param)
            messages.append({"role": "user", "content": f"Observation: {obs}"})
        return "Không hoàn thành trong giới hạn bước."

if __name__ == "__main__":
    agent = LangChainReActAgent(api_key="YOUR_OPENAI_API_KEY")

    examples = [
        "Tính 12 * 7 + 5",
        "Một hình vuông có cạnh 9. Tính chu vi và diện tích",
        "Tính (15 + 3) / 4 + 2"
    ]

    for i, q in enumerate(examples, 1):
        print(f"\n=== Ví dụ {i}: {q} ===")
        print(agent.solve(q))
```
