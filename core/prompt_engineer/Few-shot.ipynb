```python
from dotenv import load_dotenv
import os
load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate

class FewShotPromptingDemo:
    def __init__(self):
        self.llm = ChatOpenAI(model="gpt-4o-mini", api_key=OPENAI_API_KEY, temperature=0)
    
    def sentiment_analysis_demo(self, text):
        """Few-shot prompting cho phân tích sentiment"""
        
        # Định nghĩa template cho từng ví dụ
        example_template = PromptTemplate(
            input_variables=["text", "sentiment"],
            template="Text: {text}\nSentiment: {sentiment}\n"
        )
        
        # Tạo danh sách examples
        examples = [
            {
                "text": "I love this product! It's amazing and works perfectly.",
                "sentiment": "Positive"
            },
            {
                "text": "This is terrible. I hate it and want my money back.",
                "sentiment": "Negative"
            },
            {
                "text": "It's okay, nothing special but not bad either.",
                "sentiment": "Neutral"
            }
        ]
        
        # Tạo FewShotPromptTemplate
        few_shot_prompt = FewShotPromptTemplate(
            examples=examples,
            example_prompt=example_template,
            prefix="Classify the sentiment of the following text as Positive, Negative, or Neutral.\n\n",
            suffix="Text: {text}\nSentiment:",
            input_variables=["text"]
        )
        
        formatted_prompt = few_shot_prompt.format(text=text)
        return self.llm.invoke(formatted_prompt).content
    
    def code_generation_demo(self, task):
        """Few-shot prompting cho tạo code"""
        examples = [
            {
                "task": "Create a function to add two numbers",
                "code": """def add_numbers(a, b):
    \"\"\"Add two numbers and return the result.\"\"\"
    return a + b"""
            },
            {
                "task": "Create a function to check if a number is even",
                "code": """def is_even(number):
    \"\"\"Check if a number is even.\"\"\"
    return number % 2 == 0"""
            }
        ]
        
        example_template = PromptTemplate(
            input_variables=["task", "code"],
            template="Task: {task}\nCode:\n{code}\n\n"
        )
        
        few_shot_prompt = FewShotPromptTemplate(
            examples=examples,
            example_prompt=example_template,
            prefix="Here are examples of Python functions. Create a similar function for the given task.\n\n",
            suffix="Task: {task}\nCode:",
            input_variables=["task"]
        )
        
        formatted_prompt = few_shot_prompt.format(task=task)
        return self.llm.invoke(formatted_prompt).content
    
    def translation_demo(self, text, target_language):
        """Few-shot prompting cho dịch thuật"""
        examples = [
            {
                "english": "Hello, how are you?",
                "vietnamese": "Xin chào, bạn khỏe không?",
                "spanish": "Hola, ¿cómo estás?"
            },
            {
                "english": "Thank you very much.",
                "vietnamese": "Cảm ơn bạn rất nhiều.",
                "spanish": "Muchas gracias."
            }
        ]
        
        example_template = PromptTemplate(
            input_variables=["english", "vietnamese", "spanish"],
            template="English: {english}\nVietnamese: {vietnamese}\nSpanish: {spanish}\n"
        )
        
        few_shot_prompt = FewShotPromptTemplate(
            examples=examples,
            example_prompt=example_template,
            prefix="Translate the following English text to the target language:\n\n",
            suffix="English: {text}\n{target_language}:",
            input_variables=["text", "target_language"]
        )
        
        formatted_prompt = few_shot_prompt.format(text=text, target_language=target_language)
        return self.llm.invoke(formatted_prompt).content

```

```python
# Sử dụng demo
demo = FewShotPromptingDemo()

print("=== SENTIMENT ANALYSIS DEMO ===")
test_text = "The movie was decent but could have been better."
sentiment_result = demo.sentiment_analysis_demo(test_text)
print(f"Text: {test_text}")
print(f"Result: {sentiment_result}")

print("\n=== CODE GENERATION DEMO ===")
task = "Create a function to find the maximum number in a list"
code_result = demo.code_generation_demo(task)
print(f"Task: {task}")
print(f"Generated Code:\n{code_result}")

print("\n=== TRANSLATION DEMO ===")
text_to_translate = "I am learning artificial intelligence."
translation_result = demo.translation_demo(text_to_translate, "Vietnamese")
print(f"English: {text_to_translate}")
print(f"Translation: {translation_result}")
```
